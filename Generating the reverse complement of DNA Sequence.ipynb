The reverse complement of a DNA sequence is the complementary strand that joins with the main strands to form the double helix structure.

The main DNA moves from a 5 prime to 3 prime direction, while the complementary strand moves in a reverse direction, from 5 prime to 3 prime direction (making it the reverse complementary strand).

To create a complement of a DNA Sequence tou have to replace A with T, T with A, C with G and G with C
this is what produces the complement of a DNA.

```python
# Creating the complement of the DNA sequence
print("Original Sequence = ", sequence)
complement = sequence.replace("A", "t").replace("C","g").replace("T", "a").replace("G", "c")
print("Complement of the Sequence = ", complement)

# note that by writig all the replace() commands in one line and using lower case letters to  replace upper case letters, this will prevent the newly added letters from 
# being replaced when the next replace() command is running.

complement = complement.upper()
print("Complement in upper case = ", complement)
```

Creating the reverse complement of the DNA Sequence
The two colons and -1 [::-1 ] is for slicing.
the slicing format is:

sequence[start:stop:step]

in creating [::-1]

the start is not given, which prompts it to automatically start at the begining
the stop is not given, which prompts it to automatically stop at the  ending
step is -1, indicating that it should move backwards from the ending, causing a reverse sequence (or in this case a reverse complement)


```python
# creating the reverse complement of the string
reversecomplement = complement[::-1]
print("The reverse complement is = ", reversecomplement)
```

Finding the reverse complement of the DNA by defining a function.

```python
# The reverse complement of the DNA can also be done by defining a function 
#
DNA = input("please enter a DNA sequemce ")
print (DNA)

def complement (DNA):
    comp = {"A":"T", "C":"G", "G":"C", "T":"A"}
    comDNA = ''
    for b in DNA:
        comDNA += comp[b]
    return comDNA

def reverse (DNA):
    return DNA[::-1]
    
def reverse_complement (DNA):
    comDNA = complement(DNA)
    revcompDNA = reverse(comDNA)
    return revcompDNA

print ("the reverse complement is: ", reverse_complement (DNA))
```
