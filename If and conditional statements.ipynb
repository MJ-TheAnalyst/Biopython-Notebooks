IF statements are conditional statements that are used to control the flow of a program by executing specific blocks of code when conditions are met.

they act as "decision points" where program evalautes wether a condition is true or false (Boolean controlled)

Basic Syntax

if: this evalautes the intial condition, if the condition is true, the indeted block of code runs
elif: Checks additional conitions only if the previous conditions are false
else: this runs if all the pevous conditions are false
indentation:  this defines which code belongs to the if statement. incorrect indentation causes IndentationError
colon: every if, elif and else statement must end with a colon

The conditions in IF statements usually use comparison and logical  operators to evaluate data:
comparison operators like  == (equals), !=(not equals), < (less than), > (greater than) <= (Less than or equal to), >= (greater than or equal to)

logical operators like: and (means both must be true), oe (at least one must be true), not (reverses the result)

```python
# i.e 

x = 5
y = 7
if x > y:
   print ("No!")
elif x < y:
   print ("yes")
else:
   print("These two must be equal")
```

Application in bioinformatics

1). Sequence filtering: this is used to check if a DNA sequence meets a minimum length requirement before processing

Note that these codes only work when biopython library is installed so the modules like Bio.Seq can be used

2). GC content analysis: this uses the Bio.SeqUtilis module to trigger actions based on  chemical properties

```python
# Sequence filtering:
from Bio.Seq import Seq
my_seq = Seq("AGTACACTGGT")

if len(my_seq) > 10:
    print ("sequence is long enough for analysis")
```

```python
# GC content analysis
from Bio.SeqUtils import gc_fraction

gc_val = gc_fraction(my_seq)
if gc_fraction(my_seq) > 0.5:
    print("High GC_content detected")
else:
    print ("gc_content is not high")
```

While loops

```python
# Example of while loop: Counting 1 - 10
start = 0
end = 10
step = 1

counter = start
while counter < end:
    counter += step
    print(counter)
```
