Using the for loop to create a multiplication table

we use the input function to input the number whoose multiplication we want to calculate and how many multiples we want to calculate, in this example the number is 5 and the amount of multiples is 20

Using For loop, the rangee is given as (1, times +1)
then the command that foe every number in the range (1 - 20 in this example), the calculation number * 1 should be done 

```python
#Generating a table of any number to any given multiple

# Input the number whoose multiples will be calculated
number = int (input( " Submit the number whose table you need to generate"))

# Input the number of multiples that are calculated
times = int(input ("The limit to which you wish to generate the number"))

print ( "multiplcatioon table of 5")

# using for loop
for i in range (1, times +1):
    table = number * i
     
    print (number, " * ", i, " = ", table )
```

```python
# For loops on strings
DNA = "AATTGGCC"
length = len(DNA)
print(length)

for i in DNA:
    print (i)
```

```python
# For loop on string with conditional print
DNA = "AATTGGCC"

for i in DNA:
    #checking for purines
    if i == "G" or i == "A":

        print (i, " = Purine")
    else:
        print (i ," = Pyrimidine")

```

```python
# to count total purines and pyrimidines we can use the count function:

pucount = DNA.count("G" ) + DNA.count("A")
pycount = DNA.count("T" ) + DNA.count("C")
print ("The Purine count = ", pucount, " , ", "The pyrimidine count = ", pycount)

# But we cn also use the for loop

DNA = "AATTGGCC"

pucount = 0
pycount = 0

for base in DNA:
    if base == "G" or base == "A":
        pucount += 1
    else:
        pycount += 1

print ("The Purine count = ", pucount, " , ", "The pyrimidine count = ", pycount)
```

The central dogma explains how genetic information flows inside the cell and used to make proteins. It involves:

DNA - RNA (Transcription): a segement of DNA is copied into mRNA (messenger RNA) in the nucleus

RNA - Protein (Translation): the mRNA is read by ribosomes and translated in order to build proteins using amino acids


note: DNA replication is not a main part of the central dogma, it's just a process that ensures that every new cell gets it's own genetic material by making an exact copy of itself (DNA) before cell division.


```python
central_dogma = []
molecule = ["DNA","RNA","Protein"]
process = ["Replication", "Transcription", "Translation"]

for i in range (3):
    central_dogma.append(molecule[i])
    central_dogma.append(process[i])
    print(central_dogma)

print(central_dogma)
```
