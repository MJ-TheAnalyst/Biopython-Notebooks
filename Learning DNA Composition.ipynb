We rely on the user to input the DNA Sequence, we'll be using.

```python
# We take an input from the user and check the composition of the DNa Sequence and also calculate the GC content

DNA = input("Enter a DNA Sequence:")
print(DNA)
```

```python
# Making Calculations using the DNA Sequence.

# Calculating the length of input DNA Sequence
length = len(DNA)
print(length)

# Converting to upper case
DNA = DNA.upper()
print(DNA)
```

```python
# counting A,T,G,C
G = DNA.count("G")
C = DNA.count("C")
A = DNA.count("A")
T = DNA.count("T")
N = DNA.count("N")

# Printing the results
print(
    "Length = ", length, "\n"
    "N = ", N, ",",
    "G = ", G, ",",
    "C = ", C, ",",
    "A = ", A, ",",
    "T = ", T, ","


)
```

The syntax for calculating nucleotide percentge  (or in this case GC Content) is 

((nucleotide number)/total number of nucleotides in dna sequence) * 100

```python
# Calculating the percentage of nucleotides in the DNA Sequence i.e Calculating G+C % in the DNA sequence 
GC = ((G + C)/length) * 100

print  ("GC content = ", GC,"%")
```

```python
# we can also calculate the GC content by definiing the function
print(DNA)

def gccontent (DNA):
    totalC = DNA.count("C")
    totalG = DNA.count("G")
    gccontent_result = 100 * (totalC + totalG)/len(DNA)
    return gccontent_result

print ("the dna sequence has a GC content of: ", gccontent(DNA), "%")
```

```python
# You can also define your own function yourself by creating the variable
dna = "ACGTGTCACGTA"

def gccontent (dna):
    totalC = dna.count("C")
    totalG = dna.count("G")
    gccontent_result = 100 * (totalC + totalG)/len(dna)
    return gccontent_result

print ("the dna sequence has a GC content of: ", gccontent(dna), "%")
```

To calculate the DNA Melting Temperature (Tm), We use this formula for short oligonucleotides (short or small custom piecses of DNA or RNA sequences approx 15 - 50 bases):

Tm = 2 * (A + T) + 4 * (G + C)

This is called the wallace rule.

Note that more G and C content in these Oligos mean higher Tm ( because they form stronger bonds)


```python
# Calculating the Tm of the DNA Sequence
tm = (2 * (A + T) + 4 * (G + C) )
print (tm)
```

Converting DNA to RNA simply involves substituting the nucleotide T with U

```python
#Converting DNA to RNA
RNA = DNA.replace("T","U")
print ("The RNA Sequence is = ", RNA)
```
