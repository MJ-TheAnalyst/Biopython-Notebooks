In biopython sequences are often stored in Seq objects and this is ehat is used to perform the calculations and manipulation to the sequence in biopython. To store a sequence in Seq object:

```python
dna = "GCTAATCGGAAT"
rna = "GCUAAUCGGAAU"
protein = "MKWV"

from Bio.Seq import Seq
dna = Seq(dna) # The sequence can also be written as seq("GCTAATCGGAAT")

# We can even store RNA and protein sequences the same way
rna = Seq (rna)
protein = Seq(protein)
```

These sequence objects can also be used for concatenation of sequences, i.e 

```python
dna1 = "GCTAAT"
dna2 = "CGGAAT"

# to concatenate both strings:
from Bio.Seq import Seq
dna = dna1 + dna2

print(dna)

# We can even concatenate seq objects in a list

seqlist = [Seq("AGCT"), Seq("CGTA"), Seq("GATC")]
conseqlist = Seq("")
for a in seqlist:
    conseqlist += a

print (conseqlist)
```

You can see that sequence objects understand how strings work already and how sequence strings should work and automatically carries out the commands. In the same way we can also change the cases of Seq objects

```python
dna3 = "aggacta"

from Bio.Seq import Seq
print(dna3.upper())

print(dna.lower())
```

Just like strings we can also compare sequence objects using the equality operator ==

```python
from Bio.Seq import Seq

print(dna1 == dna2)
print(dna2 == dna3)
```

Something to note is that Normal Seq objects are immutable in nature and cannot be changed because of its string like nature.

To modify Seq objects we can use MutableSeq instead. i.e

```python
from Bio.Seq import MutableSeq
dna = MutableSeq(dna) # or = dna.tomutable()

dna[0] = "T"
print(dna)
```

Because of the nature of the mutable sequence, changes to the dna sequence is allowed, we can also perform other manuipulation to the sequence now like append, remove and reverse

```python
# to append 
from Bio.Seq import MutableSeq
dna4 = MutableSeq("AGGTA")

dna4.append("C")

print (dna4)


# to remove
from Bio.Seq import MutableSeq
dna4 = MutableSeq("AGGTA")

dna4.remove("G")

print (dna4)

# to reverse

from Bio.Seq import MutableSeq
dna4 = MutableSeq("AGGTA")

dna4.reverse()

print (dna4)
```

To get the complement of a dna strand, we convert A to T and C to G and vice versa, and then to get the reverse complement, we simply revserse the order of nucleotides, but in biopython these can be done in one step using the provided functions complement() and reverse_complement()

```python
dna = "GCTAATCGGAAT"

from Bio.Seq import Seq
dna = Seq(dna)
complement = dna.complement()
reversecomplement = dna.reverse_complement()

print (complement)
print (reversecomplement)
```

For DNA to RNA transcription note that actual biological transcription starts from the template strand (reverse complement) and then coverts this to mRNA i.e AGGT to TCCU (mRNA template) but in biopython the function transcribe() simply works directly with the sequence provided because MRNA can be gotten by simply switching T and U 

for simple transcription, we can do it like this random example below:

```python
dna = "GCTAATCGGAAT"
from Bio.Seq import Seq
dna = Seq(dna)

rna = dna.transcribe()

print ("RNA is:", rna)

# but when we take note of the explanation earlier:
coding_dna =Seq(dna)  # coding dna is simple just the dna in our code, the main 5' to 3' strand
mrna = coding_dna.transcribe()

print ("MRNA is:", mrna) # this is simply just like the example above

template_dna = coding_dna.reverse_complement() 
print("template strand is:", template_dna)

# in actual transcription, it begins with the creation of the reverse complement of the dna, complement of 3' to 5' dna template
# so this dna template or template_dna in this case, is the template used to begin transcription in actual biological proceses. 
# To get the mrna of dna template:

mrnatemp = template_dna.transcribe()

print("mRNA template for transcription is:", mrnatemp)

# to backtrabscribe from mRna to dna, we simply:
from Bio.Seq import Seq
Dnatempbacktranscribe = mrnatemp.back_transcribe()

print ("Backtranscript of DNA template is:", Dnatempbacktranscribe)

#  this will give back the reverse complement of DNA  since we used the template strand instead of coding dna, 
# if it was vice versa then we will get back the actual coding strand, i.e
Dnabacktranscribe = mrna.back_transcribe()

print ("Backtranscript of DNA is:", Dnabacktranscribe)

# This gives the original DNA Sequence
```

For translation of  mRNA to protein using the seq object we can simply use the translate() function.

```python
mrnatemp = Seq(mrnatemp)
translation = mrnatemp.translate()

print("The translation is:", translation)

# Note that we can also translate directly using dna
translation2 = coding_dna.translate()

print(" the translation is:", translation2)

# notice that the translation is different because it is read from different directions, the  coding strand from 5' to 3' and 
# template strand from 3' to 5 prime, to prevent that we can first create the reverse complement of the coding strand to match the template

translation2 = coding_dna.reverse_complement().translate()

print(" the translation is:", translation2)
```

by default, Biopython uses the translation tables based on the NCBI stanadrd genetic code (NCBI table 1), luckily we can specify the exact table we need for our translation. i.e suppose we rae dealing with mitochondrial sequence, we can tell biopython to use a relevant code instead:

```python
translation3 = coding_dna.reverse_complement().translate(table = "Vertebrate Mitochondrial")
print(translation3)

# to make things shorter we can also just use the table number instead of the name:
translation3 = coding_dna.reverse_complement().translate(table = 2)
print(translation3)
```

when translating codes with stop codons such as dna = "ATGGCCATTGTAATGGGCCGCTAAAGGGTGCCCGATAG"
there are so many manipulations that can be done with with i.e 

```python
# dna5 = "ATGGCCATTGTAATGGGCCGCTAAAGGGTGCCCGATAGTAC"

# or 

dna5 = "GTGAGGTAGGCATAGAGTGATTAG"
print(len(dna5))

from Bio.Seq import Seq 
dna5 = Seq(dna5)

translation4 = dna5.transcribe().translate( )

print("The translation is:", translation4)
# Notice that the result has a stop codon inside and at the end of the sequence, to stop translation at a specific stop codon, we can

translation4 = dna5.transcribe().translate( to_stop = True)

print("The translation is:", translation4)

# Notice how the stop codon is not translated or included in the code translation. We can also specify the stop symbol we want:

dna5 = "AGGTAGGCATAGAGTGATTAG"

from Bio.Seq import Seq 
dna5 = Seq(dna5)

translation4 = dna5.transcribe().translate(table = 2, stop_symbol = "#")

print("The translation is:", translation4)

# the first amino acid definition changes b/c of the change from the default table 1 to NCIB table 2
# to acknowledge a start codon which usually codes for valine but should translated to methioine, this will only happen when we 
# specify that it is a complete DNA Sequence CDS, but the sequence must satisfy the conditions of a true CDS

dna5 = "GTGAGGGCAAGTGATTAG"
from Bio.Seq import Seq 
dna5 = Seq(dna5)
translation4 = dna5.transcribe().translate( table = "Bacterial", to_stop = True, cds = True, )

print("The translation is:", translation4)
```

The traditional stop codon for coding Dna is actually ATG. However in bacterial, archaeal, plant plastid code(table 11) or vertebrate mitochondrial code(table 2) alternative start codons like GTG or TTG can be used but we must specify the actual table for the alternative to be used.

for the alternative to be acknowledged when we add a True CDS statement, we must ensure the sequence meet up with strict requirement of for valid biological coding:

1). The sequence length must be a multiple of 3

2). The sequence must end with a valid start Codon

3). The sequence must end with a valid stop codon with no stop codon in between.

Unknown Sequence Objects
this us when part of the sequence (DNA, RNA pr even protein) is missing. Python allows us store these unknown part of the sequence as sequence object. this saves an unknown sequence by denoting it as an N or X. The N( for base) or X (for amino acid) character indicates that the nucleotide exist but the exact base is unknown.  using this object a single N or X character can be used to save an antie sequence by just denoting the length of the sequence

i.e

```python
from Bio.Seq import Seq

unk = Seq("N" * 20)
print(unk)

# These can still be used for normal sequence manupulation even tho the bases is unknowm. ie
 
unkcom = unk.complement()
unkrev = unk.reverse_complement()
unktrans = unk.transcribe()
unkprotein = unk.translate
print(unkcom, unkrev, unktrans, unkprotein)
```
