SeqIO (Seqeuence input/output)
This is the main interface Biopython interface for reading and wrriting biological sequence files.It handles python files of different formats uisng a consistent and unified interface.

The core philosophy is that regardless of what format our file is in, SeqIO will always give us back SeqRecord objects
it reads, write, modifies, extract and indexes sequence files and sequence records

It has two reading functions:

SeqIO.read() for single sequence files (It must contain only one sequence),
syntax: record = SeqIO.read("filename", "format)



```python
# examples:
# for .read

from Bio import SeqIO
record = SeqIO.read("my_gene.gb", "genbank")

print(record.id)
print(len(record.seq))
print("---")
```

OR

SeqIo.Parse() for when the files have more than 1 sequence (it is used when the sequence is more than one or we don't know how many sequence are in the file and we want to process all of them). For a file with. Parsing takes exactly 2 arguments the file name or data source and the format string(always lowercaase)

syntax:
for record in SeqIO.parse("filename", "format")

Parse() returns a generator and it reads the file one record/sequence at a time rather than loading all of it in it's memory. This is because most genomic files contain millions of sequences, so to avoid the computer running out of RAM and crashing, parse reads each files, sequence by sequence (one before the other), Meaning only one record is in the memory at a time.

This mechanism is known as iterator: which allows it loop through it with the for loop and each cycle gives the nect record

```python
from Bio import SeqIO 

for record in SeqIO.parse("NC_005816.fna.txt", "fasta"):
    print(record.id)
    print(len(record.seq))
    print("---")

```

There's another type of iterator which ignores every other line on a multi-line file and prints only the first line.

next() ignores everything except the first  without warning or giving an error signal when used, so it is only best to use when you are sure that the filr has multiple sequence and you need only the first one.

in the case of uncertainty, it is best to use the read() function to avoid ignoring your data as it raises an error if there aare two or more lines in the file instead

In summary:
read() - should be used when the file has only one line or we arenn't sure if it has only one line
parse() - should be used when the file has more than one line and we want to print or process everything
next() - should be used only when we know it's a one line file and we want to only print or process the first line.

```python
from Bio import SeqIO
print_first_record = next(SeqIO.parse("NC_005816.fna.txt", "fasta"))

print(print_first_record)
```

Another operation that can be done to records using SeqIO involves converting the total records in a file into a list.

The list() function reads through the entire file and stores every record in a python list. Now the recordds behave exactly as a list so we can perform things like

- jumping to the last record
- jumping to a specific record i.e  record 50
- go backwards
- check how many records there are

WHile lists are another form of iteration, they are only best to use if the file is small enough to fit to memory, we need to know the total number of records and the user needs random access or wants to loop more than once.

```python
from Bio import SeqIO, SeqRecord
print_list_record = list(SeqIO.parse("ls_orchid.gbk (1).txt", "genbank"))

print(print_list_record)

# we ccan now perform many operations that can be done on a python list. i.e
# Check how many records
print(len(print_list_record))

# get the 1st record
first = print_list_record[0]
print(first.id)
print(first.seq)

# get the last record
last = print_list_record[-1]
print(last.id)
print(last.seq)

# get a random record i.e 5
# get the 1st record
fifth = print_list_record[4]
print(fifth.id)
print(fifth.seq)

# loop through as many times as we want
for record in print_list_record:
    print(record.id)

for record in print_list_record:
    print(len(record.seq))
```

EXTRACTING DATA USING SEQIO
specific data or informstion in sequence files like specie's name can be pulled out using to different file formats in SeqIO i.e



```python
# 1. Printing a record to see what is inside the record
from Bio import SeqIO
record_iterator = SeqIO.parse("ls_orchid.gbk (1).txt", "genbank")

first_record = next(record_iterator)
print(first_record) # This prints a human readable summary that includes ID, Name, organism, keywords and a preview of the sequence itself.
```

2. Using the Annotations 
All the metadata in the above record are stored in a pace known as record.annotations.
This is a plain python dictionary with keys and values, and each specific piece of information can be gotten using it's key. i.e


```python
# 2. Using the Annotations 

# print the whole dictionary
print(first_record.annotations)

# print only the keys (print only the available information)
print(first_record.annotations.keys())

# print just the values
print(first_record.annotations.values())

# using key to get information
print(first_record.annotations["organism"])  # Scientific latin name

print(first_record.annotations["source"])    # Common name
```

3. Extracting specific information from a genbank file
That is if you want the meta data from every record in the file, not just the first one, we can create a loop through all the records to pull the same key each time. the steps include:

- start with empty list
- loop through every record in file
- pull the value you want from each record in the nnotations dictionary
- append it to the list

```python
# 3. Extracting specific information from a genbank file

from Bio import SeqIO

all_keywords = []

for record in SeqIO.parse("ls_orchid.gbk (1).txt", "genbank"):
    all_keywords.append(record.annotations["keywords"])

print(all_keywords)

# a shorter way to do this is by using list comprehension to give a similar result with fewer lines
all_molecule_type = [record.annotations["molecule_type"]]
for record in SeqIO.parse("ls_orchid.gbk (1).txt", "genbank"):
                        print (all_molecule_type)
```

4. if extracting data from a specific position i.e the 5th record, we can do this in several ways.

a). Using the position we want to find: we can first convert the records to a list and use the index but this is only practical for small files since list() loads everything into memory. i.e

```python
from Bio import SeqIO

records = list(SeqIO.parse("ls_orchid.gbk (1).txt", "genbank"))

# to get the fifth record
fifth_record = records[4]
print(fifth_record.annotations["organism"])
```

b). Using the ID: i.e Z78530.1: Here we can loop through every record's ID until we can find it, We also need to add a break to stop the loop once we find the ID as reading the rest of the file is useless

```python
from Bio import SeqIO
target_id = "Z78530.1"

for record in SeqIO.parse("ls_orchid.gbk (1).txt", "genbank"):
    if record.id == target_id:
        print(record.annotations["organism"])
        break
```

c). Using SeqIO.index(): if  the file is large and we need fast lookup by ID we can use the index() to huikd a lookup table so we can jump directly to any record using the ID without looping. This is more efficient for large files as biopython only reads the record we ask for and leaves every other record


```python
from Bio import SeqIO
# build the index
index = SeqIO.index("ls_orchid.gbk (1).txt","genbank")

record_id = index["Z78530.1"]
print(record_id.annotations["organism"])

# we can also access another vaalue in the annotation dictionary
print(record_id.annotations["keywords"])
```

Modifying records using SeqIO: we can modify or edit attributes directly like changing Record ID or description.
it can be done by:

1. Changing attributes directly by reassigning them as you would do to any variable. this can also be done for all atrributes such as name(record.naame), Description(record.description) etc

```python
from Bio import SeqIO
record_iterator  = SeqIO.parse("ls_orchid.gbk (1).txt","genbank")
first_record1 = next(record_iterator)

print(first_record1.id)

first_record1.id = "Z78533.1"

print(first_record1.id)
```

2). Changing the Id of a FASTA file: This is done differently because of how the Id and Description of a fasta file is written, The ID and Description are written separately with a space in between (>id description). to avoid the header line looking wrong or inconsistent, when the ID is changed the description also has to be updated. i.e

```python
from Bio import SeqIO
record_iterator  = SeqIO.parse("ls_orchid.fasta.txt","fasta")
first_record1 = next(record_iterator)

print(first_record1.id)

first_record1.id = "gi|2765658|emb|Z78533.1|CIZ78533"
print(first_record1.description)
first_record1.description = first_record1.id + " " + "C.irapeanum 5.8S rRNA gene and ITS1 and ITS2 DNA"

print(first_record1.description)


```
