Tuples.
This is an inmutable list, it is declared with round brackets ().
it is iterable, i.e numbers in it can be numbered (or indexed)

```python
# Creating a tuple
numbers = (0,1,2,3,4,5)
print (numbers)
```

```python
# printing the total number of elements in a tuple

length = len(numbers)
print(length)
```

```python
# Acessing numbers in a tuple by index - zero index
print(numbers[1:])
print(numbers[2:])
print (numbers[:3])
```

The syntax for getting items from a list or tuple using index method is:

= variable[startindex:stopindex]

when theres's no start index provided, python assumes that the start index is at the beginning(position 0)
when theres's no stop index provided, python assumes that the stop index is at the ending (the very last position)

Tuples are imutable, meaning the value cannot be changed, it remains consistent throughout the program.
that is why tuples cannot be used for item assignment or be changed by assoigning it a new value

i.e numbers(2) = 5 

this will give erroes because the vakue of the tuple cannot be changed.

A tuple is a list with limited functionality.. it cannot be chnged but that can be worked around by converting it to a list using the  list() function,  making the changes and then converting back to the tuple using the tuple() function

```python
# Create a tuple
lunch = ('sandwhich', 'chips', 'soda')
print(lunch)

# Changing the content of the list
lunch[1] = 'apple'

# This will give an error messaage because a tuple does not allow for changes
```

```python
# convert the tuple to a list instead

lunch_list = list(lunch)
print(lunch_list)

# now mak your changes
lunch_list [1] = 'apple'
print (lunch_list)

# your changes  will be reflected here and you can convert it bck to a tuple using a tuple() function
lunch2 = tuple(lunch_list)
print(lunch2)
```

Tuples can also work with dictionaries, for example in creating revserse complements of DNA sequence, it can be used for base pairing of their complements. i.e

```python
bases = 'AGCT'
bases2 = bases [::-1]

print (bases)
print (bases2)

# Zip the both together to create a tuple and print the tuples by converting it to a list
zipped = zip (bases,bases2)

# note that the zip function is a one time iterator for one tme use, if you print the resukt, 
# using the zip variable again to  convert to dictionary, will not give any results as it has already been unzipped 

# Convert it to a dictionary 
bases3 = dict(zipped)
print(bases3)
```
