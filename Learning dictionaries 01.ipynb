Dictionary - this is a data structure that is indexed by key. 
It store data in a key - value pair.

This key:value pair, means a key is associated with a specific value, for a dictionaty, 
the key must be unique, but the values cn be repeated.

Dictionaries can be declared using curly brackets or using the dictionary function dict().

```python
# Declare a dictionary
marks = {"Vipin":20, "Mj":18, "Mary":19}
print(marks)

# To access a value, you can print it using the key like this:
print(marks["Mj"])
```

Dictionaries help label data learly, using unique keys.
With dictionaries, rembering positions is unneccessary, and dictionaries are easier to work with when using real life data

Dictionary items are changeable, for example

```python
# Change dictionary item using item assignment
print (marks)
marks ["Mj"] = 20
print (marks)

# Checking if keys are present in dictionaries i.e
if "Mj" in marks:
    print(1)

# Deleting values from a dictionary
del marks ["Mary"]
print(marks)

# Deleting values from a dictionary
del marks
print(marks) # This shows an error message since the dictionary has been deleted and marks by default is no longer defined
```

Values in dictionary can be a single data point or a list, A good example of this is in writing Codons,

the syntax of a dictionary is:

Name_of_dictionary  = {key:[value1, value2, value3}   for list of values

# Notes

Codons are a sequence of 3 consecutive nucleotides (trinucleotide) in DNA or mRNA that act as a unit of genetic information. 
structure: It is composed of 3 bases, eg: AUG, UUA

Each codon codes for a specific amino acid in a protein chain, they can also code the start or stop signal during translation. There are 64 possible combination of codons in the gentic code, 61 or amino acids and 3 for stop codons.

The start Codon is AUG (Methioninec) which signals the beginning of protein synthesis.
The stop codons are UAA, UAG & UGA which signals the end of protein synthesis.

Codons are specific trinucleotides in the mRNA that binds with the tRNA to begin protein synthesis, carry it out and stop protein synthesis, 

this happens when the complementary sequence of the start codons from the tRNA binds the mRNA, this signals the start of protein synthesis,
The genetic machinery (tRNA) binds to and reads mRNA in these triplets (codons) to construct proteins by linking the amino acids (synthesized by the codons) together,

WHen the complementary sequence of the stop codon binds to the stop codon on the mRNA, potein synthesis is halted.


Note that:
Redundancy in codon production enhances accuracy as multiple codons can code for the same amino acids

The role in genetics
Codons translate genetic materials into proteins.
Changes to base sequence within a codon (i.e GGT to GAT) can alter the amino acid and the resulting protein leading to mutation

```python
# Using Dictionary to write codons values in a dictionary using a list
codons = { "start": [ "AUG", "GUG"],
          "stop": [ "UAA","UAG", "UGA"],
          "Arg":["CGU","CGC","CGA","CGG","AGA", "AGG"]}

print (codons)     

# the start, stop and Arg are all keys for the corresponding list of values, Arg is short for the Amino acid, Arg.
```

Dictionaries are mutuable, values can be added and removed from a dictionary.
To add to a dictionary, you can use this syntax:
dictionary[key] = value

```python
print(len(codons))

# Adding to a dictionary
codons["Met"] = "AUG"

codons["Total_codons "] =  64
print (codons)

print(len(codons))
```

```python
# Like a list and tuple, you can iterate over a dictionary. i.e
for i in codons:
    print(i)

# Note that dictonaries only loop through keys, so it prints the keys by default, to print just the values use this syntax:
# for i in dictionary:
#     print(dictionary{i}) ; i.e

for i in codons:
    print(codons[i])

# this uses the key to find and print the values, To print both keys and values use this syntax:
# for key, value in dictionary.items():
#     print(key, value)

for key, value in codons.items():
    print(key,":", value)

```

Tuples can also work with dictionaries, for example in creating revserse complements of DNA sequence, it can be used for base pairing of their complements. i.e


```python
bases = 'AGCT'
bases2 = bases [::-1]

print (bases)
print (bases2)

# Zip the both together to create a tuple and print the tuples by converting it to a list
zipped = zip (bases,bases2)

# note that the zip function is a one time iterator for one tme use, if you print the resukt, 
# using the zip variable again to  convert to dictionary, will not give any results as it has already been unzipped 

# Convert it to a dictionary 
bases3 = dict(zipped)
print(bases3)
```
