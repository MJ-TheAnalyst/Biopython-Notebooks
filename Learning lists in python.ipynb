> A list is a collection of items stoed in a single variable. It stores multiple items in this single varaible.

**In Python:**, there are four inbuilt data types used to store a collaction of data, they include, Tuple, Set, List and Dictionary.

Some Characteristics of the List Include:

Written using square brackets: []
Can store multiple values (numbers, text, etc.)
Items are ordered and changeable
Lists are mutable or changeable ,
Some functions that can be used to alter a list include append(), insert(), delete(), extend(), pop(), remove()


```python
# Creating a list
Numbers = [0,2,1,3,5,4]
print (Numbers)


# This automatically creates a list with the 6 elements 0,2,1,3,5,4 
# We can also call a list using the list() function

numbers = list()
print(numbers)
```

```python
print (Numbers)

#Finding the number of  elements in a list
length = len(Numbers)
print("The number of elements in list = ", length)
```

```python
y = sum(Numbers)
print("The sum of numbers in the list is", y)
```

Accessing elements of a list by index
> in python indexing and counting begins at 0, making the first item in every list to have an index of 0.

```python
print(Numbers)

print(Numbers[0])
#Prints first number since indexing begins at 0

print(Numbers[1])
#Prints ssecond number since indexing begins at 0

print(Numbers[1:4])
#Prints Second to the fourth number since indexing begins at 0(between 1-5)

print(Numbers[:3])
#Prints the first number to the specified number in the list (index 0 - n-1)

print(Numbers[3:])
#Prints the fourth number to the last item in the list.
```

```python
#Sorting the list content in python in the ascending order
Numbers.sort()
print(Numbers)


#Sorting the list content in python in the descending order
Numbers.sort(reverse = True)
print(Numbers)
```

In python ther are 2 kinds of sorting methods
the In-place (.sort())
the Return a copy (.sorted())

the in-place function, rearranges all the values in the list directly and changes the orignal order of values.
> the syntax is:

`.sort()   i.e data.sort()`

the return a copy copy function takes the list and gives you back a new copy of your list.
> the syntax is:

`sorted()   i.e new = sorted (data)`

```python
# Sorting list in alphbetical order using "in-place" method 
Subjects.sort()
print(Subjects)

# Sorting list in reverse alphbetical order using reverse() function
Subjects.reverse()
print(Subjects)

#Sorting lists in alphabetical order using "Returning a copy" method
Subjects2 = sorted(Subjects)
print(Subjects2)

#Sorting lists in reverse alphabetical order using "Returning a copy" method
Subjects2 = sorted(Subjects, reverse = True)
print(Subjects2)

#note that python uses the ASCII code to check for the order of data, upper case and lower case letters 
# will be treated differently
```

#### Insert Function ####
The insert() function allows you to cut the line and put an item exactly where you want it.

> Syntax
`list.insert(index, element)`

**Index:** The numerical position where you want the new item to go (remember, Python starts counting at 0).

**Element**: the actual data you want to put there

```python
# inserting data inside a list
Subjects.insert(2,"English")
print(Subjects)
```

#### Extend() function ####
This is for adding a whole collection of things at once.

The extend() function takes another "iterable" (usually another list) and unzips it, adding each item individually to the end of your original list.


```python
# inserting a list to another list.Subjects
Subjects.extend(["communication_skills", "Soft Skills"])
print(Subjects)
```

**Removing Values From a List**

#### The pop method ####
This is a removal tool that remove some values and also hands you the item back as it kicks it out of the list.

While methods like remove() or del just delete an item and let it vanish, pop() allows you to "catch" the item so you can use it for something else (like moving a DNA sequence from a "raw" list to a "processed" list).

> The Syntax
There are two ways to use it:

**list.pop():** With no numbers inside the parentheses, it ejects the very last item in the list.

**list.pop(index):** If you provide a number (index), it ejects the item at that specific position.

```python
# Removing values from a list using pop() function
print(Subjects)

print(Subjects.pop())
# This removes the last item from Subjects, and because pop() "returns" that 
# value, the print function catches it and shows it on your screen.

print(Subjects, "\n")
# This prints the list to show that "Maths" is officially gone.
#"\n" is a "newline" character. Python to hit the "Enter" key after printing, 
# creating a blank line.

Subjects.pop(1)
print(Subjects, "\n")
#This is a targeted removal. Instead of taking the end, it takes the item at 
# Index 1. Remember, Python starts counting at 0, so Index 1 is the second item 
# ("Chemistry"). The second line the  prints the list.
```

```python
#Removing values from a list using remove() function
print(Subjects)
Subjects.remove("Physics")
print(Subjects)
```

```python
#Removing values from a list using del() function
del(Subjects[1])
print(Subjects)

#note that it removed the second subject because index is counted from 0 not 1
```
