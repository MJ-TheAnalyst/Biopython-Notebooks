```python
DNA = "GCGACTAGCTGCTAGCTACG"
print(DNA)

# in indexing, using a minus sign will start the indexing backwards, i.e

print(DNA[-0])

# This prints the very last G in the sequence. Another way to print this is using the length minus 1

print(DNA[len(DNA) - 1])

# To continue counting backwards, we always use a minus sign

print(DNA[-1])
print(DNA[-2])
print(DNA[-3]) # and so on....


```

```python
# To extract a substring, we can use the following syntax:  
# substring = string[startindex:endindex]. i.e

Sub = DNA[0:10]
print(Sub)

# you can also add a step to skip certain characters. i,e

sub = DNA[0:20:2]
print(sub)

# this skips two characters after counting the first character before counting again and so on
```
