A seqrecord is a sequence with metadata (data like ID, name and description) like we talked about in the file Using SeqRecord Objects. 

But biological sequences arent uniform: a gene (or sequence) can have coding regions, promoter regions, introns, exons and so Seqfeature is how python annoates those specific regions of interest on a sequence. 

There are three major attributes of a Seqfeature. They include:

1). .type - This is a textual description of the type of feature. in simplier terms it describes what kind of feature we're working with (eg. gene, CDS, exon, promoter).

2). .location - this tells us where on the sequence the object we're dealing with is, (it is defined by featurelocation as FeatureLocation(start, end and strand)). The Seqfeatures delegates much of it's functionality to the location object and this includes a number shortcut attributes for the property of location:
     .ref - Shorthand for .location.ref. it includes any different or reference sequence the location is refering to. Usually just none

     .ref.db - Shorthand for .location.ref_db. this specifies the datadatabase any identifier in .ref refers. Usually just none.

     .strand - shorthand for location.strand. It's the strand on the sequence that the feature is located on. For double stranded nucleotide sequnce, this may either be a 1 for the top/forward strand (5' - 3') or -1 for the bottom/Reverse complement strand or 0 if the strand is important but unknown (neutral) or none if it doesn't matter. It is none by default or single stranded sequences.


3). .qualifier - These are extra details stored as a dictionary (i.e gene name, product, locus tag). It is a literal python dictionary where the key is some kind of terse one word description of what the information contained in the value is and the value is the actual information, i.e a common key key for a qualifier might be "Evidence". and the value might be "computational(non-experimentally)" this is just to let the person know that it has not been experimentally(wet lab) confirmed. 


example: Here is a quick strand showing the anatomy of Seqfeatures




```python
from Bio.SeqRecord import SeqRecord
from Bio.SeqFeature import SeqFeature, FeatureLocation
from Bio import SeqIO

# For a sequence of at least 1000 base pairs

record = SeqIO.read("my_gene.gb", "genbank")

# Accessing features from a genbank file

for feature in record.features:
    print("Type:", feature.type)                          # eg. "gene", "CDS"
    print("Start location: ", feature.location.start)     # start position (0 based indexing)
    print("End Position:", feature.location.end)          # end position
    print("Strand:", feature.location.strand)             # +1 or -1
    print("Qualifiers:", feature.qualifiers)              # Dict of extra info


# Creating features from scratch

my_features = SeqFeature (location = FeatureLocation (100, 400, strand = 1), 
                          type = "gene",
                          qualifiers = {"gene":["BRCA1"], "NOTE":["TUMOR SUPPRESSOR"]}
                           )

print("Type2:", my_features.type)                          # eg. "gene", "CDS"
print("Start location2: ", my_features.location.start)     # start position (0 based indexing)
print("End Position2:", my_features.location.end)          # end position
print("Strand2:", my_features.location.strand)             # +1 or -1
print("Qualifiers2:", my_features.qualifiers)              # Dict of extra info
print("Type2:", my_features.type)                          # eg. "gene", "CDS"

```

Position vs Location

They help with the key idea of Seqfeatures which is to describe a region on the parent sequence for which we use a location object.

- position: refers to a single position on the sequence, which may be fuzzy or not. i.e 5, 10, <100 etc
- location: is a region of sequence bounded by some positions. i.e 5..20(5-20). Under location, we have the featurelocation and compoundlocation Objects


Featurelocation Object
For most features (especialy bacteria and simple organisms), the seqfeature is actually very simple, we just need three things:

- Where does the feature start
- Where does it end
- Which strand is it on.

This is exactly what a featurelocation objects stores.

```python
# i.e
from Bio.SeqFeature import SeqFeature, FeatureLocation
# to get a gene from position 100 - 400 on the forward strand

loc = FeatureLocation(100, 400, strand = 1)

print (loc.start)    # 100
print(loc.end)      # 400
print(loc.strand)   # 1
```

CompundLocation object - Accomodates gaps
With eukaryotes, Genes are not continous - They have exons(Coding Parts) interrupted by introns (None- coding) parts. When the cell makes a protein, the introns get spliced out and exons joined together. 

A single Featurelocation cannot represent this split structure, so biopython uses a compoudlocation, to chain multple regions together.

i.e

```python
from Bio.SeqFeature import CompoundLocation, FeatureLocation

# a gene with two exons: Position 100-300 and 500-700
exon1 = FeatureLocation(100, 300, strand = 1)
exon2 = FeatureLocation(500, 700, strand = 1)

gene_loc = CompoundLocation([exon1, exon2])

print(gene_loc)     # join{[100:300](+), [500:700](+)}. note that the + denotes that it is from a forward strand.

# Note:
# the word join in genbank files means the same thing - a feature made by joining several seperate regions togther
```

Fuzzy positions 
One complication in dealing with featurelocation comes in the position of these features themselves.
in real lab work, we dont know the exact position where a feature begins or ends, i.e we might do a dinucleotide priming experiment and discover that the start of RNA transcript starts at one of two sites. This is useful information, but the complication comes in how to represent it at the position. But to deal with this we have the position concept. We have there is 5 position classes to deal with them

1). Exact: This is used when we know exactly where the position is. This is the default when we pass a plain integer in the featurelocation.
i.e

```python
from Bio.SeqFeature import SeqFeature, FeatureLocation

loc = FeatureLocation(13, 14)  #plain integers showing exact location

print(loc.start)               # ExactPosition (13)
print(int(loc.start))          # 13
```

2). Before position: This represents a position that occurs prior to the specified site. in Gen-bank/EMBL files, it is represented as <13, signifying the real position is located somewhere less than 13. So you know the upper boundary but not the exact site

```python
from Bio.SeqFeature import SeqFeature, BeforePosition, AfterPosition

start = BeforePosition(13)  #plain integers showing exact location

print(start)               # <3
print(int(start))          # 13
```

2). After position: This represents a position that occurs after the specified site. in Gen-bank/EMBL files, it is represented as >13, signifying the real position is located somewhere more than 13. So you know the lower boundary but not the exact site


```python
from Bio.SeqFeature import SeqFeature, BeforePosition, AfterPosition

start = AfterPosition(13)  #plain integers showing exact location

print(start)               # >13
print(int(start))          # 13
```

2). Within position: This represents a position that occurs Within the specified site range i.e (1, 5), range width of 5 and lower boundary of 1. it signifies the real position is located somewhere between 1 and 5(1-4), we stop at 4, not include 5

```python
from Bio.SeqFeature import WithinPosition

startpos = WithinPosition(1, 1, 5)  # somwhere in 1-5
stoppos =  WithinPosition(20, 20, 50)

pos = FeatureLocation(start= startpos, end = stoppos, strand = 1 )

print("Feature Location string formt:")
print(pos)
print(pos.start)
print(pos.end)
print(pos.strand)
```

2). one of position: This represents a position where several possible values exist. For example, this could be used if the start codon is unclear and there are two possible start site for the gene. i.e  one_of (1, 7, 5), . it signifies the real position is located somewhere in any of the specified values.


```python
from Bio.SeqFeature import OneOfPosition, ExactPosition

pos = OneOfPosition (1, choices = [ExactPosition(1), ExactPosition(5)])  # somwhere in 1-5

print(pos)                     # one-of(1, 7, 5)
print(int(pos))                #  1 the representative integer value
```

2). unknown position: This represents a position of unknown location,  This is not used in Gen- Bank but corrresponds to the '?' feature in uniprot.

```python
from Bio.SeqFeature import UnknownPosition

pos = UnknownPosition ()  

print(pos)                     # UnknownPosition()

print(int(pos))                #  This will raise a type error as there is no numnber to extract
```

```python
# General example

from Bio import SeqFeature

# start assigning the fuzzy positions

start_pos = SeqFeature.AfterPosition(5)
# AfterPosition(5) Means: The start is somewhere after position 5

end_pos = SeqFeature.BetweenPosition(9, left = 8, right = 9)
# This means that the end is between position 8 and 9, 
# for the end position, the higher value i.e 9 here is used for slicing but not part of the range

my_location = SeqFeature.FeatureLocation (start_pos, end_pos)
# We build the location of the feature using these fuzzy positions

print(my_location)           #   [>5:(8^9)]
print(my_location.start)     #   Afterposition(5)
print(my_location.end)       #   BetweenPosition(9, left = 8, right = 9)


# To rpint just the integers (ignoring the fuzziness):
print(int(my_location.start))     #   5
print(int(my_location.end))       #   9
```
