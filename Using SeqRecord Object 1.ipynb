A SeqRecord object is a Biopython Object used to store:

- The biological Sequence, typically a Seq object
- Sequence ID, the primary ID used to identify the string
- Name, a common name or Id for the sequence Might be the same with the ID in some cases.
- Description, a human readable description of the sequence
- Additional Information like annotations

It's like the combination of a seq object and bioogical Metadata.
While normal Seq objects only store sequence data (plain text) like "AGTGATGTA", this may not be ideal for biological files containing information like  sequence, ID, description


To create a SeqRecord we just need a Seq record at  minimum
We can create a SeqRecord like in the example below

```python
# Creating the Seq object

from Bio.Seq import Seq
sequence = "AGTAGAGTTA"
sequence = Seq(sequence)

# creating the seqrecord
from Bio.SeqRecord import SeqRecord
SequenceR = SeqRecord(sequence)
```

We can also add the ID, Name and description while creating the object but if we don't, it is stored as a string that gives an "unknown" output, but this string cn be modified subsequently

i.e

```python

print(SequenceR.id)
print(SequenceR.description)
print(SequenceR.name)
print(SequenceR.annotations)
```

Notice how the properties are unknown. we can assign them to the object using the ssignment statement

i.e

```python
SequenceR.id = "AC123456"
SequenceR.name = "Human liver genome"
SequenceR.description = "I found this code online"
SequenceR.annotations ["Evidence"] = "None, Just made it up"

# to show the new properties of the sequence:

print(SequenceR.id)
print(SequenceR.description)
print(SequenceR.name)
print(SequenceR.annotations)
```

You can see that the properties have been highlighted already. The best paractice for this is to add it immediately you create the object to avoid forgetting later.

i.e:

```python
from Bio.Seq import Seq
sequence2 = "AGTAGAGTTAGT"
sequence2 = Seq(sequence2)

# creating the seqrecord
from Bio.SeqRecord import SeqRecord
Sequence2R = SeqRecord(sequence2, id = "AC1234567", name = "Human kidney genome", description = "I found this code online")


# to show the new properties of the sequence:

print(Sequence2R.id)
print(Sequence2R.description)
print(Sequence2R.name)
```

SeqRecord is heavily used when reading biological files,

for Fasta Files:
There are two ways to read a fasta file using SeqRecord.

1) Single line python file.

i.e

```python
from Bio import SeqIO

SequenceRF = SeqIO.read("NC_005816.fna.txt", "fasta")

print(SequenceRF.id)
print(SequenceRF.seq)
print(SequenceRF.name)
print(SequenceRF.description)
```

Notice how all the attrbutes are automatically assigned by the Seq Record. 

To Read a Fasta File with mutiple lines, we can use this syntax instead

This makes every sequence in the file to be treated as separate objects.

```python
from Bio import SeqIO

for record in SeqIO.parse("NC_005816.fna.txt", "fasta"):
    print(record.id)
    print(record.seq)
    print(record.name)
    print(record.description)
```

We can even save SeqRecords back into the fasta files.

for example

```python

from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO
Sequence3R = SeqRecord(Seq("AGTAGAGTTAGTAATGAT"), id = "AC1234567", name = "Human kidney genome", description = "I found this code online")

with open ("NC_005816.fna.txt","a") as output_handle:
    SeqIO.write(Sequence3R,"output_handle", "fasta")

print(Sequence3R.id)
print(Sequence3R.seq)
print(Sequence3R.name)
print(Sequence3R.description)

```

When we slice a SeqRecor, we get back a new  seqrecord where everything has been adjusted to match the lice. The features like

a). Seq - is sliced to a new region
b). features - only features fully inside the slice are kept and their cordinates shifted relatively to the new start
c). per - letter annotations 0- are slice in sync with the sequence
d). id,name,description - is carried over as it is
e). annotations - dropped because they describe the original record
f). dbdxrefs - dropped

i.e

```python
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio.SeqFeature import SeqFeature, FeatureLocation
# build seqrecord manually with 3 features
record = SeqRecord(Seq("AAAAAATGCGTAACGTTTAGCCCGATCGGCAAAGGCTAGCAAAAAA"), id = "my_gene.1", description= "test record")

#featureA: fully inside our intended slice (position 6-40)
record.features.append(SeqFeature(FeatureLocation(8,20), type = "gene", qualifiers={"gene":["gene A"]}))

#featureB: also fully inside our intended slice
record.features.append(SeqFeature(FeatureLocation(22,35), type = "CDS", qualifiers={"gene":["gene B"]}))

#featureC: fully ioutide our intended slice - will be dropped
record.features.append(SeqFeature(FeatureLocation(40,45), type = "gene", qualifiers={"gene":["gene C"]}))

print("Original length:", len(record))
print("Original features:", len (record.features))

# now slice the record
subrecord = record[6:40]

print("Sliced length:", len(subrecord))
print("Sliced features:", len (subrecord.features))
```

To reverse transcribe a SeqRecord objact we also get back a new  seqrecord where everything has been adjusted to match the lice. The features like

a). Seq - is reverse complimented
b). features - all kept and their cordinates/ strands recalculated 
c). per - letter annotations 0- kept and reversed
d). id,name,description - all dropped
e). annotations - dropped 
f). dbdxrefs - dropped

this is intentional because features can be safely recalculated but metadata like accession numbers, orgsnism names and lierature references are description of  the riginal molecules and applying them to the reverse complement would be misleading.

```python
# creating the reverse complement
rc = record.reverse_complement()
print(rc)

# notice how the records are changed, we can restore dropped attributes by passing them as arguments
rc = record.reverse_complement(id = "Testing",  # we can either pass a custom value OR
                               name = True,      # use True to copy the original attribute value
                               annotations = True,
                               dbxrefs = True,
                               )
print(rc)
print(len(rc))
print(rc.id)
print(rc.name)
print(rc.dbxrefs)
print(rc.features)
print(rc.annotations)
```
